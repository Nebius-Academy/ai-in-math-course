# PPO step by step


## 1. Policy and value models

$$
\pi^\eta(a|s) = \mathcal{N}(a|\nu^{\eta}(s),\sigma^{\eta}(s)),
\qquad
V^\theta(s).
$$


```python
import numpy as np
import torch
import torch.nn as nn
from torch.distributions import Normal


class PolicyModel(nn.Module):
    def __init__(self, state_dim, action_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(state_dim, 128), nn.ReLU(),
            nn.Linear(128, 128), nn.ReLU(),
            nn.Linear(128, 2 * action_dim), nn.Tanh(),
        )

    def forward(self, states):
        mean, log_std = self.net(states).chunk(2, dim=-1)  # two tensors (T, action_dim)
        return mean, log_std


class ValueModel(nn.Module):
    def __init__(self, state_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(state_dim, 128), nn.ReLU(),
            nn.Linear(128, 128), nn.ReLU(),
            nn.Linear(128, 1),
        )

    def forward(self, states):
        return self.net(states)  # (T, 1)
```

## 2. Advantage

The one-step advantage is
$$
A^{\theta}_t = R_t + \gamma (1 - d_t) V^{\theta}(S_{t+1}) - V^{\theta}(S_t).
$$


```python
def compute_advantages(v_model, states, rewards, next_states, dones, gamma):
    states = torch.FloatTensor(np.asarray(states, dtype=np.float32))  # list/array -> tensor (T, state_dim)
    next_states = torch.FloatTensor(np.asarray(next_states, dtype=np.float32))  # list/array -> tensor (T, state_dim)
    rewards = torch.FloatTensor(np.asarray(rewards, dtype=np.float32)).unsqueeze(1)  # (T,) -> (T, 1)
    dones = torch.FloatTensor(np.asarray(dones, dtype=np.float32)).unsqueeze(1)  # (T,) -> (T, 1)

    with torch.no_grad():
        values = v_model(states) # stop gradient through values
        next_values = v_model(next_states) # stop gradient through next_values

    return rewards + gamma * (1 - dones) * next_values - values  # (T, 1)
```

## 3. Probability ratio

PPO compares the new policy with the old one through

$$
r^{\eta}_t = \frac{\pi^\eta(A_t \mid S_t)}{\pi^{\eta_k}(A_t \mid S_t)}
= \exp\big(\log \pi^\eta(A_t \mid S_t) - \log \pi^{\eta_k}(A_t \mid S_t)\big).
$$


```python
def get_policy_distribution(pi_model, states):
    mean, log_std = pi_model(states)
    return Normal(mean, torch.exp(log_std))


def get_log_probs(pi_model, states, actions):
    dist = get_policy_distribution(pi_model, states)
    return dist.log_prob(actions).sum(dim=1, keepdim=True)


def compute_ratio(pi_model, states, actions, old_log_probs):
    new_log_probs = get_log_probs(pi_model, states, actions)
    return torch.exp(new_log_probs - old_log_probs)
```

## 4. Policy Loss: Clipped PPO objective

$$
Loss_1(\eta) = -\frac{1}{T} \sum_{t=0}^{T-1}
\min \left\{ r^{\eta} A^{\theta}_t,\; \text{clip}(r^{\eta}, 1 - \varepsilon, 1 + \varepsilon) A^{\theta}_t \right\}.
$$

The old log-probabilities and the advantages are detached because they come from the frozen trajectory under $\eta, \theta$.


```python
def ppo_policy_loss(pi_model, states, actions, old_log_probs, advantages, epsilon):
    advantages = advantages.detach()  # stop gradient through A_t
    old_log_probs = old_log_probs.detach()  # stop gradient through old policy

    ratio = compute_ratio(pi_model, states, actions, old_log_probs)

    loss_1 = ratio * advantages
    loss_2 = torch.clamp(ratio, 1 - epsilon, 1 + epsilon) * advantages

    return -torch.mean(torch.min(loss_1, loss_2))
```

## 5. Returns

Given rewards from one collected trajectory, define

$$
G_t = R_t + \gamma (1 - d_t) G_{t+1}.
$$

where $d_t = 1$ if the episode ended after step $t$.


```python
def compute_returns(rewards, dones, gamma):
    rewards = np.asarray(rewards, dtype=np.float32)  # list/array -> ndarray (T,)
    dones = np.asarray(dones, dtype=np.float32)  # list/array -> ndarray (T,)

    returns = np.zeros_like(rewards)  # allocate ndarray (T,)
    returns[-1] = rewards[-1]

    for t in range(len(rewards) - 2, -1, -1):
        returns[t] = rewards[t] + gamma * (1 - dones[t]) * returns[t + 1]

    return torch.FloatTensor(returns).unsqueeze(1)  # (T,) -> (T, 1) to match V(s)
```

## 6. Value loss

$$
Loss_2(\theta) = \frac{1}{T} \sum_{t=0}^{T-1} \left(V^\theta(S_t) - G_t\right)^2.
$$


```python
def ppo_value_loss(v_model, states, returns):
    values = v_model(states) 
    return torch.mean((values - returns) ** 2)
```

## 7. Parameter update

$$
\eta_{k+1} = \eta_k - \alpha_1 \nabla_\eta Loss_1(\eta_k),
\qquad
\theta_{k+1} = \theta_k - \alpha_2 \nabla_\theta Loss_2(\theta_k).
$$

Below is the minimal update loop, still close in style to the reference notebook.


```python
def ppo_update(
    pi_model,
    v_model,
    pi_optimizer,
    v_optimizer,
    states,
    actions,
    rewards,
    next_states,
    dones,
    gamma=0.9,
    epsilon=0.2,
    batch_size=128,
    epoch_n=30,
):
    states_np = np.asarray(states, dtype=np.float32)  # list/array -> ndarray (T, state_dim)
    actions_np = np.asarray(actions, dtype=np.float32)  # list/array -> ndarray (T, action_dim)

    states = torch.FloatTensor(states_np)  # ndarray (T, state_dim) -> tensor (T, state_dim)
    actions = torch.FloatTensor(actions_np)  # ndarray (T, action_dim) -> tensor (T, action_dim)
    returns = compute_returns(rewards, dones, gamma)  # tensor (T, 1)
    advantages = compute_advantages(v_model, states_np, rewards, next_states, dones, gamma)  # tensor (T, 1)

    with torch.no_grad():
        old_log_probs = get_log_probs(pi_model, states, actions)

    for _ in range(epoch_n):
        idxs = np.random.permutation(len(states_np))

        for i in range(0, len(states_np), batch_size):
            b_idxs = idxs[i:i + batch_size]

            b_states = states[b_idxs]  # (B, state_dim)
            b_actions = actions[b_idxs]  # (B, action_dim)
            b_returns = returns[b_idxs]  # (B, 1)
            b_advantages = advantages[b_idxs]  # (B, 1)
            b_old_log_probs = old_log_probs[b_idxs]  # (B, 1)

            pi_optimizer.zero_grad()
            pi_loss = ppo_policy_loss(
                pi_model,
                b_states,
                b_actions,
                b_old_log_probs,
                b_advantages,
                epsilon,
            )
            pi_loss.backward()
            pi_optimizer.step()

            v_optimizer.zero_grad()
            v_loss = ppo_value_loss(v_model, b_states, b_returns)
            v_loss.backward()
            v_optimizer.step()
```

## 8. Minimal PPO class

This is the same idea assembled into one compact object. The `fit(...)` method expects five arrays collected from the environment: `states`, `actions`, `rewards`, `next_states`, `dones`.


```python
class PPO(nn.Module):
    def __init__(
        self,
        state_dim,
        action_dim,
        gamma=0.9,
        batch_size=128,
        epsilon=0.2,
        epoch_n=30,
        pi_lr=1e-4,
        v_lr=5e-4,
    ):
        super().__init__()

        self.pi_model = PolicyModel(state_dim, action_dim)
        self.v_model = ValueModel(state_dim)

        self.gamma = gamma
        self.batch_size = batch_size
        self.epsilon = epsilon
        self.epoch_n = epoch_n

        self.pi_optimizer = torch.optim.Adam(self.pi_model.parameters(), lr=pi_lr)
        self.v_optimizer = torch.optim.Adam(self.v_model.parameters(), lr=v_lr)

    def get_action(self, state):
        state = torch.FloatTensor(state).unsqueeze(0)  # (state_dim,) -> (1, state_dim)
        dist = get_policy_distribution(self.pi_model, state)
        action = dist.sample()
        return action.squeeze(0).detach().numpy()  # (1, action_dim) -> (action_dim,)

    def fit(self, states, actions, rewards, next_states, dones):
        ppo_update(
            self.pi_model,
            self.v_model,
            self.pi_optimizer,
            self.v_optimizer,
            states,
            actions,
            rewards,
            next_states,
            dones,
            gamma=self.gamma,
            epsilon=self.epsilon,
            batch_size=self.batch_size,
            epoch_n=self.epoch_n,
        )
```

## 9. PPO on Pendulum

Now we train the PPO class above on `Pendulum-v1`. Data is collected episode by episode, the PPO update is made every `trajectory_n` episodes, and visualization is refreshed every 10 episodes.


```python
import numpy as np

if not hasattr(np, 'bool8'):
    np.bool8 = np.bool_  # compatibility for old gym + numpy>=2

try:
    import gymnasium as gym
except ImportError:
    try:
        import gym
    except ImportError:
        gym = None

import matplotlib.pyplot as plt
from IPython.display import HTML, clear_output, display
from matplotlib.animation import FuncAnimation


def make_pendulum_env(render=False):
    if gym is None:
        raise ImportError("Install `gymnasium[classic-control]` or `gym[classic_control]` to run the Pendulum example.")

    if render:
        try:
            return gym.make('Pendulum-v1', render_mode='rgb_array')
        except TypeError:
            env = gym.make('Pendulum-v1')
            unwrapped = getattr(env, 'unwrapped', env)
            if hasattr(unwrapped, 'render_mode'):
                unwrapped.render_mode = 'rgb_array'
            return env
    return gym.make('Pendulum-v1')


def reset_env(env):
    state = env.reset()
    if isinstance(state, tuple):
        state, _ = state
    return state


def step_env(env, action):
    step_result = env.step(action)
    if len(step_result) == 5:
        next_state, reward, terminated, truncated, _ = step_result
        done = terminated or truncated
    else:
        next_state, reward, done, _ = step_result
    return next_state, reward, done


def render_rgb_frame(env):
    try:
        frame = env.render()
        if frame is not None:
            return frame
    except Exception:
        pass

    unwrapped = getattr(env, 'unwrapped', env)
    if hasattr(unwrapped, 'render_mode'):
        try:
            unwrapped.render_mode = 'rgb_array'
            frame = env.render()
            if frame is not None:
                return frame
        except Exception:
            pass

    try:
        return env.render(mode='rgb_array')
    except TypeError:
        return None


def scale_action(action, env):
    action = np.asarray(action, dtype=np.float32)  # action -> ndarray (action_dim,)
    return np.clip(action * env.action_space.high, env.action_space.low, env.action_space.high)  # (action_dim,)


def collect_trajectory(env, agent, max_steps=200):
    states, actions, rewards, next_states, dones = [], [], [], [], []
    total_reward = 0.0
    state = reset_env(env)

    for _ in range(max_steps):
        with torch.no_grad():
            action = scale_action(agent.get_action(state), env)  # (action_dim,)

        next_state, reward, done = step_env(env, action)

        states.append(state)
        actions.append(action)
        rewards.append(reward)
        next_states.append(next_state)
        dones.append(done)

        total_reward += reward
        state = next_state

        if done:
            break

    trajectory = {
        'states': states,
        'actions': actions,
        'rewards': rewards,
        'next_states': next_states,
        'dones': dones,
    }
    return trajectory, total_reward


def collect_eval_frames(env, agent, max_steps=200):
    frames = []
    total_reward = 0.0
    state = reset_env(env)

    frame = render_rgb_frame(env)
    if frame is not None:
        frames.append(frame)

    for _ in range(max_steps):
        with torch.no_grad():
            action = scale_action(agent.get_action(state), env)  # (action_dim,)

        next_state, reward, done = step_env(env, action)
        frame = render_rgb_frame(env)
        if frame is not None:
            frames.append(frame)

        total_reward += reward
        state = next_state

        if done:
            break

    if len(frames) > 100:
        frame_step = max(1, len(frames) // 100)
        frames = frames[::frame_step]

    return frames, total_reward


def show_training_progress(frames, total_rewards, episode, episode_n, eval_reward):
    if len(frames) == 0:
        frames = [np.zeros((400, 400, 3), dtype=np.uint8)]

    fig, (ax_env, ax_reward) = plt.subplots(1, 2, figsize=(12, 4))
    fig.suptitle(f'Episode {episode}/{episode_n}, eval reward = {eval_reward:.1f}')

    image = ax_env.imshow(frames[0])
    ax_env.set_title('Pendulum every 10 episodes')
    ax_env.axis('off')

    episode_axis = np.arange(1, len(total_rewards) + 1)  # (episode,) indices for the reward curve
    ax_reward.plot(episode_axis, total_rewards, color='tab:blue')
    ax_reward.scatter([episode], [total_rewards[episode - 1]], color='tab:red', s=30)
    ax_reward.set_title('Total rewards per episode')
    ax_reward.set_xlabel('Episode')
    ax_reward.set_ylabel('Reward')
    ax_reward.set_xlim(1, episode_n)
    ax_reward.grid(True)

    def animate(frame_idx):
        image.set_data(frames[frame_idx])
        return [image]

    anim = FuncAnimation(fig, animate, frames=len(frames), interval=40, blit=True)
    html = anim.to_jshtml()
    plt.close(fig)

    clear_output(wait=True)
    display(HTML(html))
```

```python
def train_ppo_on_pendulum(
    episode_n=50,
    trajectory_n=20,
    max_steps=200,
    visualize_every=100,
):
    train_env = make_pendulum_env()
    eval_env = make_pendulum_env(render=True)

    state_dim = train_env.observation_space.shape[0]
    action_dim = train_env.action_space.shape[0]
    agent = PPO(state_dim, action_dim)

    total_rewards = []
    buffer = {
        'states': [],
        'actions': [],
        'rewards': [],
        'next_states': [],
        'dones': [],
    }

    for episode in range(1, episode_n + 1):
        trajectory, total_reward = collect_trajectory(train_env, agent, max_steps=max_steps)
        total_rewards.append(total_reward)

        for key in buffer:
            buffer[key].extend(trajectory[key])

        if episode % trajectory_n == 0 or episode == episode_n:
            agent.fit(
                buffer['states'],
                buffer['actions'],
                buffer['rewards'],
                buffer['next_states'],
                buffer['dones'],
            )
            buffer = {key: [] for key in buffer}

        if episode % visualize_every == 0 or episode == episode_n:
            frames, eval_reward = collect_eval_frames(eval_env, agent, max_steps=max_steps)
            show_training_progress(frames, total_rewards, episode, episode_n, eval_reward)

    train_env.close()
    eval_env.close()
    return agent, total_rewards


agent, total_rewards = train_ppo_on_pendulum(
    episode_n=500,
    trajectory_n=20,
    max_steps=200,
    visualize_every=50,
)
```
